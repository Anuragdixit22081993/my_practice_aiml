```python
import getpass
import whisper
import google.generativeai as genai

from google.colab import files
from IPython.display import Audio, display
from gtts import gTTS
```

```python
import os
from google.colab import userdata

os.environ["GOOGLE_API_KEY"] = userdata.get("GOOGLE_API_KEY")
genai.configure(api_key=os.environ["GOOGLE_API_KEY"])

gemini = genai.GenerativeModel("gemini-2.5-flash")


```

```python
uploaded = files.upload()
audio_file = list(uploaded.keys())[0]

print("Uploaded:", audio_file)
```

```python
whisper_model = whisper.load_model("base")

result = whisper_model.transcribe(audio_file)

text = result["text"].strip()
source_language = result["language"]

print("Detected language:", source_language)
print("Input:", text)
```

```python
languages = {
    "en": "English",
    "hi": "Hindi",

    "gu": "Gujarati",
    "pa": "Punjabi",

}

target = input("Enter target language code: ").lower()

target_name = languages.get(target, target)

print("Target language:", target_name)
```

```python
prompt = f"""
Translate this text into {target_name}.
Keep the meaning natural and conversational.
Return only the translated text.

Text:
{text}
"""

response = gemini.generate_content(prompt)

translated_text = response.text.strip()

print("Translated text:")
print(translated_text)
```

```python
output_file = "translated_voice.mp3"

tts = gTTS(
    text=translated_text,
    lang=target,
    slow=False
)

tts.save(output_file)

display(Audio(output_file))
```
