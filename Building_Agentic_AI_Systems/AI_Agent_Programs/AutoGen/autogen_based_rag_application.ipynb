```python
# ------------------------------------------------------------
# 2. Imports
# ------------------------------------------------------------

from google.colab import userdata

import chromadb

from sentence_transformers import SentenceTransformer

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
```

```python
import os
from google import genai
from google.colab import userdata

#API_KEY = put the new api key here created latest
api_key = userdata.get("GOOGLE_API_KEY")
client = genai.Client(api_key=api_key)
os.environ["GOOGLE_API_KEY"] = api_key
#import os
#from getpass import getpass

#os.environ["GOOGLE_API_KEY"] = getpass("Enter Gemini API key: ")
```

```python
# ------------------------------------------------------------
# 4. Create our small knowledge base
# ------------------------------------------------------------

documents = [

    """
    Machine learning is a branch of artificial intelligence
    that enables computers to learn patterns from data without
    being explicitly programmed. Machine learning algorithms
    perform tasks such as classification, regression,
    clustering and recommendation.
    """,

    """
    Deep learning is a subset of machine learning that uses
    artificial neural networks containing multiple layers.
    Deep learning is widely used in computer vision,
    speech recognition and natural language processing.
    """,

    """
    Generative AI refers to artificial intelligence systems
    capable of generating new content such as text, images,
    audio, video and software code. Large Language Models
    such as Gemini and GPT are examples of Generative AI.
    """,

    """
    Retrieval Augmented Generation, commonly called RAG,
    combines information retrieval with Large Language Models.

    A RAG system first retrieves relevant information from
    an external knowledge base and provides the retrieved
    information to the language model.

    The language model then uses this context to generate
    a more grounded answer.
    """,

    """
    Vector databases store numerical representations of
    information called embeddings.

    Vector databases allow semantic similarity search.

    ChromaDB, Pinecone, Weaviate, Milvus and Qdrant are
    examples of vector databases commonly used in
    Generative AI and RAG applications.
    """
]

print("✓ Knowledge base created")
print("  Documents:", len(documents))
```

```python
# ------------------------------------------------------------
# 5. Load Sentence Transformer embedding model
# ------------------------------------------------------------

print("\nLoading embedding model...")

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

print("✓ Embedding model loaded")
```

```python
# ------------------------------------------------------------
# 6. Generate embeddings for our documents
# ------------------------------------------------------------

embeddings = embedding_model.encode(documents)

print(
    "✓ Embeddings generated"
)

print(
    "  Embedding dimension:",
    len(embeddings[0])
)


# ------------------------------------------------------------
# 7. Create ChromaDB vector database
# ------------------------------------------------------------

chroma_client = chromadb.Client()

collection = chroma_client.get_or_create_collection(
    name="genai_knowledge_base"
)

ids = [
    "doc1",
    "doc2",
    "doc3",
    "doc4",
    "doc5"
]


# Avoid duplicate insertion if cell is rerun
existing = collection.get()

if len(existing["ids"]) == 0:

    collection.add(
        ids=ids,
        documents=documents,
        embeddings=embeddings.tolist()
    )

print("✓ Documents stored in ChromaDB")
```

```python
# ------------------------------------------------------------
# 8. RAG retrieval function
# ------------------------------------------------------------

def retrieve_documents(query: str, n_results: int = 2) -> str:

    """
    Retrieve documents relevant to the user's question
    from ChromaDB.
    """

    # Convert question into embedding
    query_embedding = embedding_model.encode(query)

    # Search vector database
    results = collection.query(
        query_embeddings=[
            query_embedding.tolist()
        ],
        n_results=n_results
    )

    retrieved_documents = results["documents"][0]

    # Combine retrieved documents
    context = "\n\n".join(
        retrieved_documents
    )

    return context

```

```python
# ------------------------------------------------------------
# 9. Create AutoGen RAG Tool
# ------------------------------------------------------------

async def search_knowledge_base(query: str) -> str:

    """
    Search the internal knowledge base using semantic
    vector similarity and return relevant information.
    """

    print(
        "\n[RAG TOOL] Searching knowledge base for:",
        query
    )

    context = retrieve_documents(
        query,
        n_results=2
    )

    print(
        "[RAG TOOL] Relevant information retrieved."
    )

    return context


# ------------------------------------------------------------
# 10. Configure Gemini model for AutoGen
# ------------------------------------------------------------

model_client = OpenAIChatCompletionClient(

    model="gemini-2.5-flash",

    api_key=api_key,

    base_url=
    "https://generativelanguage.googleapis.com/v1beta/openai/",

    model_info={
        "vision": True,
        "function_calling": True,
        "json_output": True,
        "family": "unknown",
        "structured_output": True
    }
)

print("✓ Gemini model configured")
```

```python


# ------------------------------------------------------------
# 11. Create AutoGen RAG Agent
# ------------------------------------------------------------

rag_agent = AssistantAgent(

    name="RAG_Assistant",

    model_client=model_client,

    tools=[
        search_knowledge_base
    ],

    system_message="""

    You are an intelligent RAG assistant.

    You have access to a tool called
    search_knowledge_base.

    For questions related to:

    - Artificial Intelligence
    - Machine Learning
    - Deep Learning
    - Generative AI
    - RAG
    - Vector Databases

    ALWAYS search the knowledge base first.

    Use the retrieved information to answer
    the user's question.

    If the knowledge base does not contain
    enough information, clearly say that.

    Give clear and concise explanations.

    """
)

print("✓ AutoGen RAG Agent created")
```

```python
# ------------------------------------------------------------
# 12. Ask a question
# ------------------------------------------------------------

question = """
What is Retrieval Augmented Generation
and why are vector databases useful in RAG?
"""

print("\n" + "=" * 60)
print("USER QUESTION")
print("=" * 60)

print(question)


# ------------------------------------------------------------
# 13. Run AutoGen Agent
# ------------------------------------------------------------

result = await rag_agent.run(
    task=question
)


# ------------------------------------------------------------
# 14. Print final answer
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("AUTOGEN + GEMINI RAG ANSWER")
print("=" * 60)

print(
    result.messages[-1].content
)


# ------------------------------------------------------------
# 15. Close model client
# ------------------------------------------------------------

await model_client.close()
```
